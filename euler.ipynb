Problem 2

```python
# make fibo sequence
fibo = [1,2]
for i in range(1,10000):
    new_item = fibo[-1] + fibo[-2]
    fibo.append(new_item)
    if new_item > 4000000:
        break

# sum even terms
my_sum = 0
for item in fibo:
    if item % 2 == 0:
        my_sum += item

print(my_sum)
```
